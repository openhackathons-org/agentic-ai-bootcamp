## Learning Objectives

- Setup Opencode with Nemotron 3 Nano as the coding assistant

Configure the local endpoint set up in [01_inference_endpoint](./01_inference_endpoint.ipynb)

```python
%%bash
cat > ../../opencode.json <<EOF
{
  "\$schema": "https://opencode.ai/config.json",
  "provider": {
    "lm-studio": {
      "npm": "@ai-sdk/openai-compatible",
      "name": "Local NIM",
      "options": {
        "baseURL": "${INF_URL}"
      },
      "models": {
        "nvidia/nemotron-3-nano": {
          "name": "nemo-3-nano"
        }
      }
    }
  }
}
EOF
```

Open a new terminal in jupyter lab by clicking on File -> New -> Terminal

<div style="text-align: center;">
  <img src="images/jupyterlab-new-terminal.png" style="width: 900px; height: auto;">
</div>

In the new terminal window, input `opencode` and press enter. This will get you into the opencode session. If `opencode` is not installed, install it using `curl -fsSL https://opencode.ai/install | bash`.

<div style="text-align: center;">
  <img src="images/opencode.png" style="width: 900px; height: auto;">
</div>

Input `/models` and select `nemo-3-nano`. You can also choose any other model available in the catalog.

Opencode is now setup. You can use this to ask any question on the tutorials or assist you in the challenges.

Example prompt: `Explain the code in challenge/mcp-servers/invoice/src/mcp_server_invoice/server_http.py`
